## 1. Basic Recursion Patterns (1-10)

Basic recursion forms the foundation for more complex patterns. These examples demonstrate:
- Simple linear recursion with a base case
- Multiple recursive calls (tree recursion)
- Parameter transformation during recursion
- Building results from recursive calls

```python
# Example 1: Factorial - Simple linear recursion
print("Factorial(5):", factorial(5))
print("Pattern: Linear recursion with base case")

# Visualize the recursion tree
def visualize_factorial(n, indent=0):
    if n <= 1:
        print("  " * indent + f"factorial({n}) = 1")
        return 1
    print("  " * indent + f"factorial({n}) = {n} * factorial({n-1})")
    result = n * visualize_factorial(n-1, indent+1)
    print("  " * indent + f"= {result}")
    return result

print("\nRecursion tree for factorial(4):")
visualize_factorial(4)
```

```python
# Example 2: Fibonacci - Multiple recursive calls
print("Fibonacci(5):", fibonacci(5))
print("Pattern: Binary tree recursion (inefficient without memoization)")

# Visualize overlapping subproblems
def fib_calls(n, memo=None):
    if memo is None:
        memo = {}
    if n <= 1:
        return 1
    if n in memo:
        return memo[n]
    
    calls = 1 + fib_calls(n-1, memo) + fib_calls(n-2, memo)
    memo[n] = calls
    return calls

print("\nNumber of function calls for fibonacci(10):", fib_calls(10))
print("This demonstrates exponential growth without memoization!")
```

## 2. Tree Recursion Patterns (11-20)

Tree recursion is fundamental for LeetCode problems involving:
- Binary tree traversals

```python
# Create a sample tree for demonstrations
#       1
#     /   \
#    2     3
#   / \
#  4   5
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)

# Example 11: Inorder traversal
print("Inorder traversal:", inorder_traversal(root))
print("Pattern: Left, Process, Right")

# Example 12: Preorder traversal
print("Preorder traversal:", preorder_traversal(root))
print("Pattern: Process, Left, Right")

# Example 13: Postorder traversal
print("Postorder traversal:", postorder_traversal(root))
print("Pattern: Left, Right, Process")
```

```python
# Example 14: Max depth - Max of recursive results
print("Max depth:", max_depth(root))
print("Pattern: 1 + max(left_depth, right_depth)")

# Visualize depth calculation
def visualize_depth(node, depth=0):
    if not node:
        return
    print("  " * depth + f"Node {node.val} at depth {depth}")
    visualize_depth(node.left, depth + 1)
    visualize_depth(node.right, depth + 1)

print("\nTree structure:")
visualize_depth(root)
```

## 3. Backtracking Patterns (21-30)

Backtracking is crucial for combinatorial problems:
- Generate all combinations/permutations
- Constraint satisfaction
- Path finding with constraints
- Subset generation

```python
# Example 21: Generate Parentheses
print("Generate parentheses(3):")
for combo in generate_parentheses(3):
    print(combo)
print("\nPattern: Track open/close counts, add '(' if open < n, ')' if close < open")

# Visualize the decision tree
def visualize_parentheses(n, s="", left=0, right=0, depth=0):
    if len(s) == 2 * n:
        print("  " * depth + s)
        return
    
    if left < n:
        print("  " * depth + f"Add '(' -> {s + '('}")
        visualize_parentheses(n, s + '(', left + 1, right, depth + 1)
    
    if right < left:
        print("  " * depth + f"Add ')' -> {s + ')'}")
        visualize_parentheses(n, s + ')', left, right + 1, depth + 1)

print("\nDecision tree for generate_parentheses(2):")
visualize_parentheses(2)
```

```python
# Example 22: Subsets - Include/exclude pattern
nums = [1, 2, 3]
print(f"Subsets of {nums}:")
for subset in subsets(nums):
    print(subset)
print("\nPattern: For each element, either include it or exclude it")

# Example 23: Permutations - Track used elements
print(f"\nPermutations of {nums}:")
for perm in permutations(nums):
    print(perm)
print("\nPattern: Track which elements have been used, build permutation step by step")
```

## 5. Dynamic Programming with Recursion (41-50)

DP with recursion combines recursive thinking with memoization:
- Overlapping subproblems
- Optimal substructure
- State caching
- Bottom-up vs top-down approaches

```python
# Example 41: Climbing Stairs - Fibonacci with memoization
print("Climbing stairs(5):", climb_stairs(5))
print("Pattern: f(n) = f(n-1) + f(n-2) with memoization")

# Compare with and without memoization
import time

def fib_no_memo(n):
    if n <= 1:
        return n
    return fib_no_memo(n-1) + fib_no_memo(n-2)

start = time.time()
result_no_memo = fib_no_memo(35)
time_no_memo = time.time() - start

start = time.time()
result_memo = climb_stairs(35)
time_memo = time.time() - start

print(f"\nFibonacci(35) without memo: {result_no_memo} (time: {time_no_memo:.4f}s)")
print(f"Fibonacci(35) with memo: {result_memo} (time: {time_memo:.4f}s)")
print(f"Speedup: {time_no_memo/time_memo:.1f}x faster with memoization!")
```

```python
# Example 42: House Robber - Choice-based DP
nums = [1, 2, 3, 1]
print(f"House robber({nums}): {rob(nums)}")
print("Pattern: At each house, choose to rob or skip based on max value")

# Visualize the decision process
def visualize_rob(nums, i=0, current=0):
    if i >= len(nums):
        print(f"  " * i + f"End: total = {current}")
        return current
    
    print(f"  " * i + f"House {i} (value: {nums[i]})")
    
    # Rob current house
    print(f"  " * i + f"-> Rob house {i}: {current} + {nums[i]} = {current + nums[i]}")
    rob_current = visualize_rob(nums, i + 2, current + nums[i])
    
    # Skip current house
    print(f"  " * i + f"-> Skip house {i}: {current}")
    skip_current = visualize_rob(nums, i + 1, current)
    
    return max(rob_current, skip_current)

print("\nDecision process for rob([1,2,3]):")
visualize_rob([1, 2, 3])
```

## Pattern Recognition Guide

### When to Use Basic Recursion:
- Problem can be broken down into smaller similar subproblems
- There's a clear base case
- Each recursive call moves toward the base case

### When to Use Tree Recursion:
- Working with tree data structures
- Need to explore multiple paths from each node
- Problem involves hierarchical relationships

### When to Use Backtracking:
- Need to find all possible solutions
- Problem involves making a sequence of choices
- Need to explore and abandon paths that don't work

### When to Use Divide and Conquer:
- Problem can be divided into independent subproblems
- Solutions to subproblems can be combined
- Often involves sorting or searching

### When to Use DP with Recursion:
- Problem has overlapping subproblems
- Optimal substructure exists
- Same subproblem is solved multiple times

### Key Recursive Patterns to Remember:
1. **Linear**: Process one element, recurse on rest
2. **Binary**: Make two recursive calls (left/right, include/exclude)
3. **Multi-way**: Make multiple recursive calls from each state
4. **Accumulator**: Pass along a running total or result
5. **Helper Function**: Use additional parameters for state
6. **Memoization**: Cache results of expensive recursive calls
